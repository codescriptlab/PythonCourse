# **Series**

```python
import pandas as pd
import numpy as np

# 1. Creando una Series (Unidimensional) con Índices Personalizados
heroes = ['Iron Man', 'Batman', 'Spider-Man', 'Wonder Woman', 'Deadpool']
universos = ['Marvel', 'DC', 'Marvel', 'DC', 'Marvel']

series_heroes = pd.Series(data=universos, index=heroes, name="Universo_Pertenece")

print("--- NUESTRA PRIMERA SERIES DE PANDAS ---")
print(series_heroes)
print(f"\n¿Qué universo tiene Batman?: {series_heroes['Batman']}")
```

```python
import pandas as pd
import numpy as np

# Imaginemos el daño infligido por los héroes en dos batallas diferentes
daño_batalla_1 = pd.Series([100, 85, 90], index=['Iron Man', 'Batman', 'Thor'])
daño_batalla_2 = pd.Series([50, 95, 40], index=['Thor', 'Iron Man', 'Spider-Man'])

print("--- ⚔️ DAÑO BATALLA 1 ---")
print(daño_batalla_1)
print("\n--- ⚔️ DAÑO BATALLA 2 (En diferente orden y con Spider-Man) ---")
print(daño_batalla_2)

# SUMA DE SERIES: Verás la magia de la alineación
daño_total = daño_batalla_1 + daño_batalla_2

print("\n--- 📈 DAÑO TOTAL COMBINADO ---")
print(daño_total)   # OJO CON LAS SUMA DE NUMERO + NADA !!!!!!
```

```python
# Solución profesional: Rellenar con 0 si el héroe no estuvo en esa batalla
daño_total_limpio = daño_batalla_1.add(daño_batalla_2, fill_value=0)

print("\n--- 🧼 DAÑO TOTAL LIMPIO (Sin valores NaN) ---")
print(daño_total_limpio)
```

```python
print("\n--- 📊 ESTADÍSTICAS RÁPIDAS DEL DAÑO ---")
print(f"Suma total de daño infligido por todos: {daño_total_limpio.sum()} HP")
print(f"Promedio de daño por superhéroe: {daño_total_limpio.mean():.2f} HP")
print(f"Máximo daño registrado en un solo héroe: {daño_total_limpio.max()} HP")

# El truco intermedio: ¿Quién fue el héroe que hizo ese daño máximo?
# idxmax() - Te dice el valor 195 pero tambien la clave asociada en este caso IRON MAN
heroe_maximo = daño_total_limpio.idxmax()
print(f"🏆 El héroe más destructor fue: {heroe_maximo} con {daño_total_limpio[heroe_maximo]} de daño")
```

# **DataFrame**

```python
# 2. Creando un DataFrame (Bidimensional) desde un Diccionario
datos_completos = {
    'Heroe': ['Iron Man', 'Batman', 'Spider-Man', 'Wonder Woman', 'Deadpool'],
    'Universo': ['Marvel', 'DC', 'Marvel', 'DC', 'Marvel'],
    'Poder_Ataque': [95, 88, 92, 96, 85],
    'Presupuesto_Millones': [250.5, 300.0, 120.0, 150.0, 58.3]
}

df_heroes = pd.DataFrame(datos_completos)

print("\n--- NUESTRO PRIMER DATAFRAME COMPLETO ---")
print(df_heroes)
```

# **Crear Columnas Derivadas (Operaciones Vectorizadas)**

```python
# Creamos una columna nueva operando dos columnas existentes
df_heroes['Costo_Por_Poder'] = df_heroes['Presupuesto_Millones'] / df_heroes['Poder_Ataque']

print("\n--- 💸 DATAFRAME CON NUEVA MÉTRICA DE COSTO ---")
print(df_heroes[['Heroe', 'Poder_Ataque', 'Presupuesto_Millones', 'Costo_Por_Poder']])
```

# **Filtros Avanzados con Condiciones Múltiples**

```python
# Aplicamos una máscara booleana con dos condiciones entre paréntesis ()
# Es obligatorio usar los paréntesis para que Pandas no se confunda
filtro_rentables = (df_heroes['Universo'] == 'Marvel') & (df_heroes['Presupuesto_Millones'] < 150)

# Pasamos el filtro al DataFrame
heroes_eficientes = df_heroes[filtro_rentables]

print("\n--- ⚡ HÉROES DE MARVEL CON PRESUPUESTO < 150M ---")
print(heroes_eficientes)
```

# **Agrupacion de datos (GroupBy)**

```python
df_heroes.describe()
```

```python
# Agrupamos por la columna 'Universo' y calculamos la media aritmética de las columnas numéricas
reporte_universos = df_heroes.groupby('Universo')[['Poder_Ataque', 'Presupuesto_Millones']].mean()

print("\n--- ⚔️ MARVEL VS DC: ANÁLISIS DE PROMEDIOS ---")
print(reporte_universos)
```
